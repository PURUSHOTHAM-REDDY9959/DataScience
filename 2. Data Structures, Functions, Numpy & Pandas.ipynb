# Section C: Exploring Pandas

#### 11.	Create a small DataFrame manually with 10 rows and columns: 'Name', 'Age', 'City', and 'Salary'. Then:
- Use .info(), .describe()
- Select 'Name' and 'City' columns
- Drop 'City' column
- Fill any missing values in 'Salary' column with the mean
- Remove any duplicate rows


```python
data = {
    "Name": ["Ram", "Akhil", "Mahesh", "Midhun", "Teja",
             "Dileep", "Madhan", "Raj", "Tej", "Bhanu"],
    "Age": [23, 25, 22, 24, 26,
            28, 27, 24, 29, 25],
    "City": ["Delhi", "Mumbai", "Chennai", "Delhi", "Pune",
             "Hyderabad", "Kolkata", "Mumbai", "Delhi", "Mumbai"],
    "Salary": [50000, 60000, np.nan, 55000, 65000,
               70000, np.nan, 62000, 72000, 60000]
}

df = pd.DataFrame(data)
df
```

```python
df.info()
```

```python
df.describe()
```

```python
df[["Name", "City"]]
```

```python
df_drop = df.drop(columns=["City"])
df_drop
```

```python
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
df
```

```python
df_no_duplicates = df.drop_duplicates()
df_no_duplicates
```
