# EXPLORATORY DATA ANALYSIS ON A DATASET

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Cardiotocographic.csv")
df.head()
```

#### 1.	Data Cleaning and Preparation:

```python
df.info()
```

```python
df.shape
```

```python
df.columns
```

```python
df.isnull().sum()
```

```python
df.fillna(df.mean(numeric_only=True), inplace=True)
```

```python
df.dtypes
```

```python
#outliers
plt.figure(figsize=(15,8))
sns.boxplot(data=df)
plt.xticks(rotation=90)
plt.show()
```

#### Statistical Summary

```python
df.describe()
```

```python
#mean
df.mean(numeric_only=True)
```

```python
#median
df.median(numeric_only=True)
```

```python
#Standard Devaition
df.std(numeric_only=True)
```

```python
Q1 = df.quantile(0.25)
Q3 = df.quantile(0.75)

IQR = Q3 - Q1

IQR #Inter Quartile Range
```

- The numerical variables have different ranges and variability.
- Some variables contain outliers, as observed from the boxplots.
- Mean and median differ for a few variables, indicating skewed distributions.

#### 3.	Data Visualization:


```python
df.hist(figsize=(15,12))
plt.tight_layout()
plt.show()
```

## Box Plots

```python
plt.figure(figsize=(15,8))
sns.boxplot(data=df)
plt.xticks(rotation=90)
plt.show()
```

## Correlation Heatmap

```python
plt.figure(figsize=(12,8))
sns.heatmap(df.corr(), annot=True, cmap="coolwarm")
plt.show()
```

## Scatter Plot

```python
plt.figure(figsize=(6,4))
sns.scatterplot(x='LB', y='ASTV', data=df)
plt.show()
```

# Pair Plot

```python
sns.pairplot(df[['LB','AC','UC','ASTV','NSP']])
plt.show()
```

## Violin Plot

```python

plt.figure(figsize=(8,5))
sns.violinplot(y=df['LB'])
plt.show()
```
