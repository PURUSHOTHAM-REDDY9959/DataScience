# Exercise 1: Prime Numbers

```python
number=int(input("Enter a Number:"))
if number>1:
    
    for i in range(2,number):
        if number%i==0:
            print("NOT PRIME")
            break
    else:
        print(f"{number} is a Prime Number ")
else:
    print(number,"NOT PRIME")
```

# Exercise 4: Hospital Billing System


```python
days=int(input("Enter No of Days stayed:"))
lab_test=int(input("Enter No of Lab Tests done:"))
medicine_charges=int(input("Enter medical charges:"))

total_bill=(2000*days)+(1500)+(300*lab_test)+medicine_charges
print()
print(f"Total Bill = ₹{total_bill}")

if total_bill>10000:
    final_bill=total_bill-(0.1*total_bill)
print(f"Final Bill on discounting = ₹{final_bill}")
```
